**Q2. Records count**

```python
import pandas as pd

# Download the dataset
!wget https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv

# Read the dataset with Pandas
df = pd.read_csv('car_fuel_efficiency_2026.csv')

# Display the first few rows of the DataFrame
print(df.head())
```

```python
num_records = df.shape[0]
print(f"The dataset contains {num_records} records.")
```

**Q3. Fuel types**

```python
num_fuel_types = df['fuel_type'].nunique()
print(f"There are {num_fuel_types} distinct fuel types in the dataset.")
```

**Q4. Missing values**

```python
# Count columns with missing values
missing_cols = df.isnull().any().sum()
print(f"Number of columns with missing values: {missing_cols}")

# Also show which columns specifically have missing values and how many
print(df.isnull().sum())
```

**Q5. Max fuel efficiency**

```python
# Filter for cars from Asia and find the maximum fuel efficiency
max_asia_fuel_efficiency = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
print(f"The maximum fuel efficiency of cars from Asia is: {max_asia_fuel_efficiency}")
```

**Q6. Median value of horsepower**

```python
# 1. Calculate the initial median value of the horsepower column
initial_median = df['horsepower'].median()
print(f"Initial median value of horsepower: {initial_median}")

# 2. Calculate the most frequent value (mode) of the horsepower column
most_frequent_horsepower = df['horsepower'].mode()[0]
print(f"Most frequent value of horsepower: {most_frequent_horsepower}")

# 3. Fill missing values in horsepower with the most frequent value
df['horsepower'] = df['horsepower'].fillna(most_frequent_horsepower)

# 4. Calculate the median value of horsepower once again
new_median = df['horsepower'].median()
print(f"New median value of horsepower after filling missing values: {new_median}")
```

**Q7. Sum of weights**

```python
import numpy as np

# 1. Select all the cars from Asia
asia_cars = df[df['origin'] == 'Asia']

# 2. Select only columns vehicle_weight and model_year
asia_subset = asia_cars[['vehicle_weight', 'model_year']]

# 3. Select the first 7 values
asia_subset_7 = asia_subset.head(7)

# 4. Get the underlying NumPy array. Let's call it X
X = asia_subset_7.values
print("X:")
print(X)

# 5. Compute matrix-matrix multiplication between the transpose of X and X
XTX = X.T.dot(X)
print("\nXTX:")
print(XTX)

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)
print("\nXTX inverse:")
print(XTX_inv)

# 7. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200]
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Multiply the inverse of XTX with the transpose of X, and then multiply by y
w = XTX_inv.dot(X.T).dot(y)
print("\nw:")
print(w)

# 9. What's the sum of all the elements of the result?
sum_w = w.sum()
print(f"\nSum of all elements of w: {sum_w}")
```
